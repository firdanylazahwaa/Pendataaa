## Preprocessing

**1. Pengertian Preprocessing**

Preprocessing data merupakan tahapan awal dalam proses data mining maupun machine learning yang bertujuan untuk menyiapkan data sebelum dilakukan analisis. Data yang diperoleh biasanya masih berupa data mentah (raw data) yang belum siap digunakan secara langsung. Dalam kondisi tersebut, sering ditemukan beberapa masalah seperti:

- Adanya data yang kosong (missing value)

- Data yang tidak konsisten

- Perbedaan skala antar atribut

- Data yang mengandung kesalahan atau noise

Oleh karena itu, preprocessing dilakukan untuk memperbaiki kualitas data agar lebih terstruktur dan layak digunakan, sehingga hasil analisis menjadi lebih optimal.

**2. Tahapan Preprocessing Data**

Dalam preprocessing, terdapat beberapa langkah penting yang umumnya dilakukan, yaitu:

- Data Cleaning

  Tahap ini berfokus pada pembersihan data dari kesalahan, data duplikat, maupun nilai yang tidak lengkap.

- Data Integration

  Merupakan proses penggabungan data dari berbagai sumber menjadi satu kesatuan dataset.

- Data Transformation

  Pada tahap ini, data diubah ke dalam format yang sesuai dengan kebutuhan analisis.

- Data Normalization

  Proses penyesuaian nilai data ke dalam skala yang sama agar tidak terjadi dominasi oleh atribut tertentu.

**3. Normalisasi Data**

Normalisasi adalah proses mengubah nilai data ke dalam rentang tertentu tanpa mengubah pola atau hubungan antar data tersebut.

Tujuan normalisasi:

- Menyamakan skala antar atribut

- Menghindari pengaruh berlebih dari nilai yang besar

- Mempermudah proses perhitungan algoritma seperti KNN, WKNN, dan clustering

- Beberapa metode normalisasi yang sering digunakan antara lain:

- Min-Max Normalization

- Z-Score Normalization

- Decimal Scaling

**3.1** Min-Max Normalization

Metode ini mengubah nilai data ke dalam rentang tertentu, biasanya antara 0 hingga 1.

Rumus:

$$
X' = \frac{X - X_{\min}}{X_{\max} - X_{\min}}
$$

Keterangan:

$$
\begin{aligned}
X &= \text{nilai asli} \\
X_{\min} &= \text{nilai minimum} \\
X_{\max} &= \text{nilai maksimum} \\
X' &= \text{hasil normalisasi}
\end{aligned}
$$

Contoh:

Data IPK:

2, 3, 4

Nilai minimum = 2
Nilai maksimum = 4

Normalisasi nilai 3:

$$
\frac{4 - 2}{3 - 2} = 0.5
$$

Hasil:

| Nilai Asli | Normalisasi |
|------------|-------------|
| 2          | 0           |
| 3          | 0.5         |
| 4          | 1           |


**3.2** Z-Score Normalization

Z-Score digunakan untuk mengubah data berdasarkan nilai rata-rata dan standar deviasi.

$$
Z = \frac{X - \mu}{\sigma}
$$

Keterangan:

$X$ = nilai data  
$\mu$ = rata-rata (mean)  
$\sigma$ = standar deviasi  
$Z$ = hasil normalisasi  

Contoh:

Data IPK:

2, 3, 4

Rata-rata:

$$
\frac{3 + 2 + 3 + 4}{4} = 3
$$

Jika nilai yang dinormalisasi adalah 3:

$$
Z = 0
$$

Artinya nilai tersebut berada tepat di rata-rata data.

**3.3** Decimal Scaling Normalization

Metode ini dilakukan dengan cara membagi nilai data dengan pangkat 10 berdasarkan jumlah digit terbesar.

Rumus:

$$
x' = \frac{x}{10^j}
$$

Keterangan:

$X$ = nilai data  
$j$ = jumlah digit terbesar  


Contoh:

Data PO:

200000, 300000, 400000

Nilai terbesar = 400000 → memiliki 6 digit

Maka:

$$
\frac{200000}{10^6} = 0.2 \\
\frac{300000}{10^6} = 0.3 \\
\frac{400000}{10^6} = 0.4
$$

Hasil:

| Nilai Asli | Normalisasi |
|------------|------------|
| 200000     | 0.2        |
| 300000     | 0.3        |
| 400000     | 0.4        |


**4. Kesimpulan**

Preprocessing merupakan tahap penting dalam pengolahan data sebelum dilakukan analisis atau pemodelan. Salah satu bagian utama dari preprocessing adalah normalisasi data yang bertujuan untuk menyamakan skala antar atribut.

Metode normalisasi yang umum digunakan meliputi:

- Min-Max untuk rentang 0–1

- Z-Score berdasarkan rata-rata dan standar deviasi

- Decimal Scaling menggunakan pembagian pangkat 10

Dengan melakukan preprocessing yang baik, data menjadi lebih siap untuk dianalisis sehingga hasil yang diperoleh lebih akurat dan dapat diandalkan.
