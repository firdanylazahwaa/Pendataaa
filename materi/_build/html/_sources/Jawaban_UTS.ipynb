image.png

Gambar berikut menunjukkan alur proses analisis data menggunakan Orange dari tahap input data hingga evaluasi model.

Pada penelitian ini digunakan metode K-Nearest Neighbors (KNN) buat mengklasifikasikan data tanah menjadi dua kategori, yaitu Subur dan Tidak Subur.

KNN bekerja dengan cara:

- Data baru akan dibandingkan dengan data yang sudah ada.
- Dicari beberapa tetangga terdekat (misalnya K = 5).
- Kelas ditentukan dari mayoritas tetangga tersebut.

**2. Lakukan Pemrosesan Data tersebut**

Dataset memiliki missing values dan data kategorikal, sehingga dilakukan beberapa tahap pemrosesan data.

  a. Select Columns

  Memilih fitur yang digunakan dalam analisis dan menentukan label sebagai target klasifikasi.

  image.png

  b. Imputasi Data

  Karena ada data yang kosong (missing value), maka perlu diisi:

- Data angka → diisi dengan rata-rata
- Data kategori → diisi dengan nilai yang paling sering muncul

  Tujuannya supaya data bisa diproses tanpa error.

  image.png

  c. Transformasi Data (Continuize)

  Fitur Tekstur Tanah yang awalnya berupa kategori diubah jadi angka, Karena KNN cuma bisa membaca data numerik.

  image.png

  d. Normalisasi Data

  Data dinormalisasi supaya semua fitur punya skala yang sama,karena KNN menghitung jarak. Kalau tidak dinormalisasi, nilai besar bisa lebih berpengaruh.

  image.png

  **3. Hitung Metrik Evaluasi**

  Evaluasi model dilakukan menggunakan beberapa metrik berikut:

  a. Accuracy

  $$
  Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
  $$

  Accuracy menunjukkan persentase jumlah prediksi yang benar dari seluruh data.

  b. Precision

  $$
  Precision = \frac{TP}{TP + FP}
  $$

  Precision menunjukkan ketepatan model dalam memprediksi kelas Subur.

  c. Recall

  $$
  Recall = \frac{TP}{TP + FN}
  $$

  Recall menunjukkan kemampuan model dalam mendeteksi seluruh data Subur.

  d. F1-Score

  $$
  F1 = \frac{2 \times Precision \times Recall}{Precision + Recall}
  $$

  F1-Score merupakan rata-rata harmonik antara precision dan recall.
