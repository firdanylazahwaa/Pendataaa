## Analisis Model Explainability Skforecast

*1. Analisa prediksi tentang apa?*

Model digunakan untuk memprediksi nilai Demand (permintaan energi/listrik) pada periode berikutnya berdasarkan data historis Demand dan variabel eksternal berupa Temperature (suhu).

*2. Bagaimana bentuk data trainingnya?*

Data training terdiri dari:

*Input (X_train)*
- lag_1
- lag_2
- lag_3
- lag_4
- lag_5
- lag_6
- lag_7
- Temperature

Lag merupakan nilai Demand pada periode-periode sebelumnya.

*Output (y_train)*
- Demand yang akan diprediksi pada periode berikutnya.

Jadi, input model adalah data Demand masa lalu dan Temperature, sedangkan outputnya adalah nilai Demand yang diprediksi.

*3. Apa itu lag?*

Lag adalah data historis atau data masa lalu yang digunakan sebagai fitur untuk melakukan prediksi.

Contoh:
- lag_1 = Demand 1 periode sebelumnya
- lag_2 = Demand 2 periode sebelumnya
- lag_3 = Demand 3 periode sebelumnya

Pada kasus ini digunakan lag_1 sampai lag_7.

*4. Jelaskan proses analisis yang dilakukan*

1. Mengumpulkan data Demand dan Temperature.
2. Membuat fitur lag dari data Demand.
3. Membagi data menjadi training dan testing.
4. Melatih model forecasting menggunakan data training.
5. Melakukan prediksi pada data testing.
6. Menggunakan metode SHAP untuk menjelaskan hasil prediksi.
7. Menganalisis fitur yang paling berpengaruh terhadap prediksi.
8. Menarik kesimpulan berdasarkan kontribusi masing-masing fitur.
