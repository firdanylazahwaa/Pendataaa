## Pengukuran Jarak

**1. Pengertian Pengukuran Jarak**

Pengukuran jarak (distance measurement) adalah teknik untuk menghitung tingkat kemiripan atau perbedaan antara dua objek data. Dalam data mining, pengukuran jarak sangat penting karena digunakan pada:

* Clustering (misalnya K-Means)

* K-Nearest Neighbor (KNN)

* Similarity search

* Data preprocessing


**Tujuan Pengukuran Jarak**

Pengukuran jarak bertujuan untuk:

- Mengetahui kemiripan antar data

- Mengelompokkan data yang mirip

- Membantu proses klasifikasi

- Mengidentifikasi outlier

**Jenis Tipe Data dalam Pengukuran Jarak**

Sebelum menghitung jarak, kita harus tahu tipe datanya.

1. Data Numerik

Data berupa angka yang bisa dihitung langsung.

Contoh:

* Umur

* Tinggi badan

* Harga

* Bisa pakai Euclidean, Manhattan, dll.

2. Data Kategorikal (Nominal)

Data berupa label tanpa urutan.

Contoh:

* Warna (Merah, Biru)

* Jenis kelamin

* Kota

* Biasanya pakai matching / encoding

3. Data Ordinal

Data kategori yang punya urutan.

Contoh:

* Tingkat kepuasan (Rendah, Sedang, Tinggi)

* Ranking

* Harus diubah ke angka sesuai urutan

4. Data Campuran (Mixed Data)

Data yang berisi kombinasi:

* numerik

- kategorikal

- ordinal

- Untuk data campuran biasanya digunakan Gower Distance

**Metode Pengukuran Jarak yang Umum**

1. Euclidean Distance

  * Rumus:

   $$
   d(x,y) = \sqrt{\sum_{i=1}^{n}(x_i - y_i)^2}
   $$

  Ciri:

- Untuk data numerik

- Paling populer

- Digunakan di K-Means

2. Manhattan Distance

Rumus:

$$
d(x,y) = \sum_{i=1}^{n} \left| x_i - y_i \right|
$$

Ciri:

- Untuk data numerik

- Lebih tahan terhadap outlier

3. Hamming Distance

Digunakan untuk:

* data kategorikal

* data biner

Cara kerja:

- hitung jumlah atribut yang berbeda

4. Gower Distance

Digunakan untuk:

* data campuran
* numerik
* kategorikal
* ordinal

Cara Kerja Gower Distance

Langkah umum:

- Normalisasi data numerik

- Konversi ordinal ke ranking

- Bandingkan kategorikal (match/tidak)

- Hitung rata-rata jarak semua atribut

- Nilai hasil:

0 = sangat mirip

1 = sangat berbeda

Dataset terdiri dari 1000 data siswa dengan beberapa atribut yang menggambarkan kondisi akademik dan latar belakang siswa.

* Setiap baris = 1 siswa

* Setiap kolom = 1 atribut

Dataset ini termasuk data campuran (mixed attributes) karena mengandung:

* Nominal

* Binary

* Ordinal

* Numerik

**DAFTAR ATRIBUT DAN TIPENYA**

Berikut klasifikasi lengkapnya:


| Atribut                         | Tipe Data               | Keterangan                |
|----------------------------------|--------------------------|----------------------------|
| gender                           | Binary (symmetric)       | male / female              |
| race/ethnicity                   | Nominal                  | group A–E                  |
| parental level of education      | Ordinal                  | tingkatan pendidikan       |
| lunch                            | Binary (asymmetric)      | standard / free-reduced    |
| test preparation course          | Binary (asymmetric)      | completed / none           |
| math score                       | Numerik                  | 0–100                      |
| reading score                    | Numerik                  | 0–100                      |
| writing score                    | Numerik                  | 0–100                      |


**Penjelasan Tiap Jenis Atribut**
**A. Nominal**

Contoh:

- race/ethnicity

Tidak ada urutan.

Pengukuran jarak:

- dij = 0 jika sama
- dij = 1 jika berbeda

(Simple Matching)

**B. Binary**

Binary Symmetric

Contoh:

- gender

Karena male dan female sama pentingnya.

- Rumus:

Simple Matching

- Binary Asymmetric

Contoh:

- test preparation course

- lunch

Karena nilai "completed" lebih bermakna daripada "none"

Biasanya pakai:

Jaccard Coefficient

**C. Ordinal**

Contoh:

- parental level of education

Urutan:

- high school
- some college
- associate's degree
- bachelor's degree
- master's degree

Langkah :

- Ubah ke ranking

- Normalisasi ke [0,1]

- Hitung seperti numerik

**D. Numerik**

Contoh:

- math score

- reading score

- writing score

Bisa pakai:

- Manhattan (L1)

- Euclidean (L2)

**3. Implementasi Data Campuran Dengan (Orange)**

image.png

Setelah dilakukan perhitungan jarak secara manual, tahap selanjutnya adalah menghitung jarak secara komputasional menggunakan **Orange Data Mining**. Widget *Distances* menghasilkan matriks berukuran **1000 × 1000**, sesuai dengan jumlah objek pada dataset *Students Performance*.

Setiap baris dan kolom merepresentasikan satu siswa, sehingga setiap sel pada matriks menunjukkan **nilai jarak antara dua siswa** berdasarkan atribut numerik (math score, reading score, writing score) yang sebelumnya telah dipilih.

Pada tampilan matriks, nilai jarak divisualisasikan menggunakan gradasi warna hijau:

- Warna **lebih muda** → jarak kecil → siswa memiliki performa yang mirip.  
- Warna **lebih gelap** → jarak besar → perbedaan performa semakin tinggi.  

Diagonal utama bernilai **0**, karena objek dibandingkan dengan dirinya sendiri. Nilai di luar diagonal menunjukkan jarak yang dihitung menggunakan **Manhattan Distance**.

Matriks ini menjadi dasar untuk tahap berikutnya, yaitu proses *Hierarchical Clustering*, untuk mengamati pola pengelompokan siswa berdasarkan kemiripan nilai akademiknya.
